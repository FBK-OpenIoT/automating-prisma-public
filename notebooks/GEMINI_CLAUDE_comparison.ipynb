# DOI comparison between Gemini and Claude review files


```python
from pathlib import Path

import pandas as pd


def find_project_root() -> Path:
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / "data" / "results" / "review_final.csv").exists():
            return candidate
    raise FileNotFoundError("Could not find data/results/review_final.csv from the current working directory.")


root = find_project_root()
gemini_path = root / "data" / "results" / "review_final.csv"
claude_path = root / "data" / "results" / "CLAUDE_review_final.csv"

gemini_df = pd.read_csv(gemini_path)
claude_df = pd.read_csv(claude_path)

def normalized_dois(df: pd.DataFrame) -> set[str]:
    if "DOI" not in df.columns:
        raise KeyError(f"Missing DOI column. Available columns: {list(df.columns)}")
    return set(
        df["DOI"]
        .dropna()
        .astype(str)
        .str.strip()
        .str.lower()
    )


gemini_dois = normalized_dois(gemini_df)
claude_dois = normalized_dois(claude_df)

common_dois = sorted(gemini_dois & claude_dois)
gemini_only_dois = sorted(gemini_dois - claude_dois)
claude_only_dois = sorted(claude_dois - gemini_dois)

def print_doi_list(title: str, dois: list[str]) -> None:
    print(f"\n{title} ({len(dois)}):")
    if not dois:
        print("  None")
        return
    for doi in dois:
        print(f"  - {doi}")


print(f"Gemini DOI count: {len(gemini_dois)}")
print(f"Claude DOI count: {len(claude_dois)}")

print_doi_list("DOIs in common", common_dois)
print_doi_list("DOIs only in Gemini review_final.csv", gemini_only_dois)
print_doi_list("DOIs only in Claude CLAUDE_review_final.csv", claude_only_dois)
```
